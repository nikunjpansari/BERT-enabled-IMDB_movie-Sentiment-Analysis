```python
import numpy as np
import tensorflow as tf
```

```python
# -------------------------------
# 1. One-Hot Encoding of Sentences
# -------------------------------

# Define sample sentences.
sentences = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]

# Define the vocabulary size for one-hot encoding.
vocab_size = 10000
```

```python
# Generate one-hot representations for each sentence.
one_hot_repr = [tf.keras.preprocessing.text.one_hot(sentence, vocab_size) for sentence in sentences]
print("One-hot encoded representations:")
for idx, rep in enumerate(one_hot_repr):
    print(f"Sentence {idx+1}: {rep}")
```

```python
# -------------------------------
# 2. Padding the Sequences
# -------------------------------
# Define the fixed sequence length.
sent_length = 8

# Pad the one-hot encoded sequences so that all have the same length.
# Using pre-padding so that shorter sequences have zeros on the left.
from tensorflow.keras.preprocessing.sequence import pad_sequences
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_length)
print("\nPadded sequences (embedded_docs):")
print(embedded_docs)
```

```python
# -------------------------------
# 3. Building the Embedding Model
# -------------------------------
# Define the dimensionality of the embedding.
embedding_dim = 10

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding

# Create a simple Sequential model with an Embedding layer.
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=sent_length))
# For a simple demonstration, compile with mean squared error.
model.compile(optimizer='adam', loss='mse')

# Display the model's architecture.
model.summary()
```

```python
# -------------------------------
# 4. Predicting Embedding Representations
# -------------------------------
# Obtain predictions (i.e. embedded representations) for all padded sequences.
predictions = model.predict(embedded_docs)
print("\nPredictions (embeddings) for all documents:")
print(predictions)

# Predict for a single document.
single_prediction = model.predict(embedded_docs[0:1])
print("\nPrediction (embedding) for the first document:")
print(single_prediction)
```
